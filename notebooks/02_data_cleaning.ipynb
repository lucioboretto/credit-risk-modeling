```python
# Import the libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/Users/lucioboretto/Desktop/Data-Science-Portfolio/credit-risk/data/hmeq.csv")
```

```python
df_clean = df.copy()

df_clean.shape
```

```python
df_clean.isnull().sum().sort_values(ascending=False)
```

```python
missing_summary = pd.DataFrame({
    'missing': df_clean.isnull().sum(),
    'missing_pct': (df_clean.isnull().sum() / len(df_clean) * 100).round(2),
    'dtype': df_clean.dtypes
})

missing_summary = missing_summary[
    missing_summary['missing'] > 0
].sort_values('missing_pct', ascending=False)

missing_summary
```

```python
numeric_missing_cols = [
    'DEBTINC',
    'DEROG',
    'DELINQ',
    'MORTDUE',
    'YOJ',
    'NINQ',
    'CLAGE',
    'CLNO',
    'VALUE'
]

for col in numeric_missing_cols:
    df_clean[f'{col}_missing'] = df_clean[col].isna().astype(int)
```

```python
df_clean.sample(10)
```

```python
categorical_missing_cols = ['JOB', 'REASON']

for col in categorical_missing_cols:
    df_clean[col] = df_clean[col].fillna('Unknown')
```

```python
df_clean.to_csv('../data/HMEQ_clean.csv', index=False)
```
